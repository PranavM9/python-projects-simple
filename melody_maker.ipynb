```python
def compose_song():
  num_mel = int(input("Enter the number of melodies: "))
  len_mel = int(input("Enter the length of each melody: "))
  melodyLst = []

  for i in range(1, num_mel+1):
    melody = []
    print(f"Composing melody #{i}")
    for j in range(1, len_mel+1):
      note = input(f"Enter note #{j}: ")
      melody.append(note)
    melodyLst.append(melody)

  return melodyLst

compose_song()
```

```python
NOTES = "CDEFGAB"

def most_common_naturals(melodyList):
  most_common_list = []
  for i in range(len(melodyList)):
    curr_melody = melodyList[i]
    most_common_list.append(find_common(curr_melody))

  return most_common_list

def find_common(curr_melody):
  most_common = ""
  most_repeats = 0
  for i in range(len(curr_melody)):
    curr_note = curr_melody[i]
    repeats = num_repeats(curr_note, curr_melody)

    if most_repeats < repeats and NOTES.find(curr_note) != -1:
      most_repeats = repeats
      most_common = curr_note

    elif most_repeats == repeats and NOTES.find(curr_note) != -1:
      if NOTES.index(curr_note) < NOTES.index(most_common):
        most_common = curr_note

  return most_common


def num_repeats(curr_note, curr_melody):
  repeats = 0
  for i in range(len(curr_melody)):
    if curr_note == curr_melody[i]:
      repeats+=1

  return repeats

def main():
  print(most_common_naturals([["A", "B", "C"], ["C♯", "C", "A"]]))

if __name__ == "__main__":
  main()
```
