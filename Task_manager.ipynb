```python
def print_todo(todo):
  print("Today's TODOs:")
  if len(todo) == 0:
    print("  You have nothing to do yet today! Relax!")
  else:
    for i in range(len(todo)):
      task = todo[i]
      print(f"{i+1}. {task}")

def add_item(todo):
  task = input("What would you like to add? ")
  if len(todo) == 0:
    todo.append(task)
  else:
    index = input(f"Where in the list should it be (1-{len(todo)+1})? (Enter for end): ")
    if index == '':
      todo.append(task)
    else:
      index = int(index) - 1
      todo.insert(index, task)

def mark_item(todo):
  index = int(input(f"Which item did you complete (1-{len(todo)})? ")) - 1
  todo.pop(index)

def save_tasks(todo):
  file_name = input("Enter file name: ")
  with open(file_name, "w") as f:
    for task in todo:
      f.write(task+"\n")

def load_tasks(file_name, todo):
  with open(file_name, "r") as f:
    for line in f:
      todo.append(line.split()[0])

def mark_item_range(todo):
  start_index = int(input(f"Which item did you complete first (1-{len(todo)})? ")) - 1
  end_index = int(input(f"Which item did you complete last (1-{len(todo)})? "))
  for i in range(start_index, end_index):
    todo.pop(start_index)



def main():
  EXTENSION_FLAG = True
  INPUTS = 'AMLSQ'
  todo = []

  console = ""

  print("Welcome to your TODO List Manager!")

  while console.upper() != 'Q':
    print("What would you like to do?")
    console = input("(A)dd TODO, (M)ark TODO as done, (L)oad TODOs, (S)ave TODOs, (Q)uit? ")
    if console.upper() == 'A':
      add_item(todo)
    elif console.upper() == 'M':
      if EXTENSION_FLAG:
        mark_item_range(todo)
      else:
        mark_item(todo)
    elif console.upper() == 'L':
      file_name = input("Enter file name: ")
      load_tasks(file_name, todo)
    elif console.upper() == 'S':
      save_tasks(todo)
    elif INPUTS.find(console.upper()) == -1:
      print(f"Unknown input: {console}")
    print_todo(todo)


if __name__=='__main__':
  main()
```
