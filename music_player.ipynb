```python
"""
music_player.py

This program allows the user to interact with the PLayer object by
providing an interface that prompts the user to select an option
of their choice correspoding to adding or playing a song, printing
user playlist or history, clearing user history and deleting
from history.

"""
def menu():
  """
  Prints out menu options.

  """
  print("(A) Add song")
  print("(P) Play song")
  print("(H) Print history")
  print("(V) View playlist")
  print("(C) Clear history")
  print("(D) Delete from history")
  print("(Q) Quit")
  print()
  print()

def main():
  user = Player()
  print("Welcome to the CSE 122 Music Playlist!")
  console = ""

  while console.upper() != "Q":
    menu()
    console = input("Enter your choice: ")

    if console.upper() == 'A':
      song = input("Enter song name: ")
      if user.add_song(song):
        print(f"Successfully added {song}")
    elif console.upper() == 'P':
      print(f"Playing song: {user.play()}")
    elif console.upper() == 'H':
      user.print_hist()
    elif console.upper() == 'V':
      user.print_playlist()
    elif console.upper() == 'C':
      user.clear_hist()
    elif console.upper() == 'D':
      print("A positive number will delete from recent history.")
      print("A negative number will delete from the beginning of history.")
      num_songs = int(input("Enter number of songs to delete: "))
      user.delete_history(num_songs)
    print()
    print()

if __name__ == "__main__":
  main()
```

```python
from collections import deque
#class comment
class Player():
  """
  This class allows the client to perform functions on the user's
    playlist and history of songs.

  Attributes:
      __playlist (deque): Songs in playlist.
      __history (list): Played songs in reverse chronological order.

  """

  def __init__(self):
    """
    Initializes an empty playlist and history.

    """
    self.__playlist: deque[str] = deque([]) # queue
    self.__history: list[str] = [] # stack

  def add_song(self, song: str) -> bool:
    """
    Adds the provided song to the user's playlist.

    Args:
        song: Name of song to be added.

    Returns:
        bool: true if song was added false if not.

    """
    self.__playlist.appendleft(song)
    return True

  def play(self) -> str:
    """
    Plays the song at the top of the playlist, the one that
    was added first. Adds the played song to user history.

    Returns:
        str: Name of song that was played.

    """
    song = self.__playlist.pop()
    self.__history.append(song)
    return song

  def print_hist(self) -> None:
    """
    Prints user history in reverse chronological order,
    most recent song appears on top.

    """
    temp_stack = []
    #Prints song in history and transfers it to auxillary storage
    size = len(self.__history)
    for i in range(size):
      song = self.__history.pop()
      print(song)
      temp_stack.append(song)

    #Restores original state of history stack
    size = len(temp_stack)
    for i in range(size):
      self.__history.append(temp_stack.pop())

  def print_playlist(self) -> None:
    """
    Prints the user playlist, song that was added first
    appears on top.

    """
    size = len(self.__playlist)
    for i in range(size):
      song = self.__playlist.pop()
      print(song)
      self.__playlist.appendleft(song)

  def clear_hist(self) -> None:
    """
    Clears user history.

    """
    self.__history.clear()

  def delete_history(self, num_songs: int) -> None:
    """
    Deletes the given number of songs from user history.
    Deletes from recent history if given number is
    positive and from beginning of the history if it is
    negative.

    Args:
        num_songs: Number of songs to be deleted.

    Raises:
        ValueError: if abs(num_songs) exceeds history length.


    """
    if abs(num_songs) > len(self.__history):
      raise ValueError("number of songs to delete is greater than size of history")
    temp_stack = []
    if num_songs < 0:
      while len(self.__history) != 0:
        temp_stack.append(self.__history.pop())
      for i in range(abs(num_songs)):
        temp_stack.pop()
      while len(temp_stack) != 0:
        self.__history.append(temp_stack.pop())
    else:
      for i in range(num_songs):
        self.__history.pop()

```
