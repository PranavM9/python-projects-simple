```python
def intro_message(user):
  print(f"There are {user.num_stocks()} on the market: ")
  ticker_lst = user.get_stocks()
  price_lst = user.get_prices()
  for i in range(user.num_stocks()):
    print(f"{ticker_lst[i]}: {price_lst[i]} ")

def main():
  file_name = input("Enter stocks file name: ")
  user = Portfolio(file_name)
  print("Welcome to the CSE 122 Stocks Simulator!")
  intro_message(user)

  console = ""
  while console.upper() != "Q":
    print("Menu: (B)uy, (Se)ll, (D)isplay, (S)ave, (Q)uit")
    console = input("Enter your choice: ")
    if console.upper() == "B":
      ticker = input("Enter the stock ticker: ")
      budget = float(input("Enter your budget: "))
      user.buy(ticker, budget)
    elif console.upper() == "D":
      user.display_portfolio()
    elif console.upper() == "Se":
      stock_name = input("Enter the stock ticker: ")
      num_shares = float(input("Enter number of shares to sell: "))
      user.sell(stock_name, num_shares)
    elif console.upper() == "S":
      new_file_name = input("Enter new portfolio file name(.txt): ")
      user.save(new_file_name)
  print(f"Your portfolio is currently valued at: ${user.valuation()}")

if __name__ == "__main__":
  main()

```

```python
class Portfolio:
  def __init__(self, file_name):
    self.__stock_tickers = []
    self.__prices = []
    self.__owned = []
    self.__num_stocks = 0

    f = open(file_name, "r")
    self.__num_stocks = int(f.readline())
    f.readline()
    for i in range(self.__num_stocks):
      line = f.readline()
      word_lst = line.split()
      self.__stock_tickers.append(word_lst[0])
      self.__prices.append(float(word_lst[1]))
      self.__owned.append(0)



  def num_stocks(self):
    return self.__num_stocks

  def get_stocks(self):
    return self.__stock_tickers[:]

  def get_prices(self):
    return self.__prices[:]

  def buy(self, ticker, budget):
    if self.__stock_tickers.count(ticker) == 0:
      raise Exception("Mentioned stock dosen't exist.")
    index = self.__stock_tickers.index(ticker)
    price = self.__prices[index]
    num_shares_bought = float(budget/price)
    self.__owned[index] = num_shares_bought

  def display_portfolio(self):
    print("Portfolio: ")
    for share in self.__owned:
      if share != 0:
        ticker = self.__stock_tickers[self.__owned.index(share)]
        print(f"{ticker} {share}")

  def sell(self, ticker, num_shares):
    if self.__stock_tickers.count(ticker) == 0:
      raise Exception("Mentioned stock dosen't exist.")
    index = self.__stock_tickers.index(ticker)
    num_owned = self.__owned[index]
    if num_owned > num_shares:
      self.__owned[index] -= num_shares
    else:
      raise Exception("Number of shares being sold is more than owned")

  def save(self, new_file_name):
    with open(new_file_name, "w") as f:
      f.write("Portfolio: \n")
      for share in self.__owned:
        if share != 0:
          ticker = self.__stock_tickers[self.__owned.index(share)]
          f.write(f"{ticker} {share} \n")

  def valuation(self):
    valuation = 0.0
    for share in self.__owned:
      index = self.__owned.index(share)
      share_price = self.__prices[index]
      valuation += float(share*share_price)
    return valuation

```
